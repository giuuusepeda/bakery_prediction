## Multiple Linear Regression (with Interactions and Harmonic Features)
---

Now we will change the regression equation. We are dealing with sales data across multiple product categories. This is a *Time Series* where each product might have its own seasonal/ periodic sales pattern. A simple linear term might not catch the full effect. So we will do more feature engineering and perform multiple linear regression with interaction among variables and harmonic Features.

We have already added some **Calender Features** by including the *KielerWoche, Feiertag and Ferien* flags.We will add one more flag called *is_weekend* to seperate the weekdays and weekends. And we will model the *cyclic effect* or seasonlaity of months and weeks by creating two **Fourier Terms** for each them. Where the terms have the following forms -

$$ FT_{sine} = sin(\frac{2\pi*t}{T}) $$
$$ FT_{cosine} = cos(\frac{2\pi*t}{T}) $$

and $t = 1, 2, ... T$ and $T$ stands for the period length.

We will also introduce the so called **Interaction Effects** - where multiple variables interact with each other. As each member of the product groups have different sales pattern across time, we will let each of them interact with the independent variables. Our argument is - the sales of each product category (Warengruppe) depends on *Temperature* and if the current date is a **special day** (weekend/ public holiday/ school hoilday/ kieler woche). We will then include the Fourier Terms for the Week and Month .

Our regression equation will now be of this form -

$$ y = Global\space Intercept + (Coefficents \times Global\space Seasonality\space Variables) $$
$$ +  (Product\space Groups\space \times Interaction\space Variables) $$


### Preprocessing

- adding weekend flag (0/1)
- adding the **Fourier Terms** (sine and cosine) (like x,y co-ordinate) of Week and Month

```python
# Preprocessing Function
def preprocess(df):
    """ Preprocess the input DataFrame by adding cyclical features and weekend indicator."""

    df = df.copy()
    df['Datum'] = pd.to_datetime(df['Datum'])
    
    # Create is_weekend: 1=Monday...7=Sunday based on 2013-07-01 being Monday and having Wochentag=1
    df['is_weekend'] = df['Wochentag'].apply(lambda x: 1 if x >= 6 else 0)
    
    # Cyclical Features for Month (period=12)
    df['sin_Monat'] = np.sin(2 * np.pi * df['Monat'] / 12)
    df['cos_Monat'] = np.cos(2 * np.pi * df['Monat'] / 12)
    
    # Cyclical Features for Wochentag (period=7)
    df['sin_Wochentag'] = np.sin(2 * np.pi * df['Wochentag'] / 7)
    df['cos_Wochentag'] = np.cos(2 * np.pi * df['Wochentag'] / 7)
    
    # Jahreszeit (Season) - Not needed now
    
    return df



# Apply preprocessing to all datasets
train_df = preprocess(train_df)
val_df = preprocess(val_df)
test_df = preprocess(test_df)
```

### Performing the Extended Linear Regression

Let's build the model.

```python
# Define Formula
# Metric : We want to maximize Adjusted R-squared. 
# Strategy: Allow Warengruppe (Product Group) to interact with key drivers.
# Hypothesis:
# 1. Sales patterns differ by product (C(Warengruppe)).
# 2. Weekend effect differs by product (e.g., Bread vs Cakes).
# 3. KielerWoche effect differs by product.
# 4. Temperature effect differs by product.
# 5. Assuming Global Seasonality for month and week.

formula = (
    "Umsatz ~ C(Warengruppe) * (KielerWoche + is_weekend + Temperatur + Ferien + Feiertag) "
    "+ sin_Monat + cos_Monat + sin_Wochentag + cos_Wochentag"
)

# Fit Model
model = smf.ols(formula=formula, data=train_df).fit()

# Print Summary to check R-squared and Adj. R-squared
print(model.summary())

```

### Calculating the Performnace Metrics on Training and Validation Dataset

Writing a function to evaluate the following performance metrics for our model - MAE, MSE, RMSE, MAPE.
We will also evaluate MAPE for diffrent product groups, for both Training and Validation dataset. We are building container dataframes to hold the metrics and then using seaborn to make plots.

```python
# writing performance evaluation function
def evaluate_model(model, df, dataset_name="Dataset"):
    """
    Evaluate a regression model and report performance metrics.

    This function calculates global metrics (R2, MAE, MSE, RMSE, MAPE) and 
    provides a breakdown of the Mean Absolute Percentage Error (MAPE) 
    per 'Warengruppe'.

    Dependencies
    ------------
    - pandas
    - numpy
    - sklearn.metrics

    Parameters
    ----------
    model : object
        The trained regression model that follows the Scikit-Learn 
        estimator API (must have a .predict() method).
    df : pandas.DataFrame
        The dataset containing the features and the target column 'Umsatz'. 
        Must also include a 'Warengruppe' column for grouping.
    dataset_name : str, optional
        Label for the dataset being evaluated, used in the printed 
        report heading. Defaults to "Dataset".

    Returns
    -------
    y_pred : numpy.ndarray
        The predicted values for the entire input dataframe.
    metrics : tuple
        A tuple containing the global performance metrics in the order:
        (R2 Score, MAE, MSE, RMSE, MAPE).
    warengruppes : list 
        The unique group identifiers found in the 'Warengruppe' column.
    warengruppes_mape : list of float
        The MAPE values calculated for each corresponding group in 
        `warengruppes`.
    
    """
    y_pred  = model.predict(df)             #Predicted values
    y_true = df['Umsatz']                   #True values    

    #Global Metrics
    r2_score_value = r2_score(y_true, y_pred)          #R² score
    mae = mean_absolute_error(y_true, y_pred)         #mean absolute error
    mse = mean_squared_error(y_true, y_pred)         #mean squared error 
    rmse = np.sqrt(mse)                          #root mean squared error
    mape = mean_absolute_percentage_error(y_true, y_pred) #mean absolute percentage error

    # create a tuple of the metrics
    metrics = (r2_score_value, mae, mse, rmse, mape) 
   
    print("#" * 30)
    print(f"{dataset_name} Performance:")
    print("#" * 30)
    
    print(f"R² Score: {r2_score_value:.4f}")
    print(f"MAE: {mae:.4f}")
    print(f"MSE: {mse:.4f}")
    print(f"RMSE: {rmse:.4f}")
    print(f"MAPE: {mape:.4%}") #shows as percentage
    print("-" * 30)

    # Group-level MAPE breakdown
    warengruppes = df['Warengruppe'].unique()
    warengruppes_mape = []
   
    for wg in warengruppes:
        df_wg = df[df['Warengruppe'] == wg]
        y_pred_wg = model.predict(df_wg)
        y_true_wg = df_wg['Umsatz']
        mape_wg = mean_absolute_percentage_error(y_true_wg, y_pred_wg)
        warengruppes_mape.append(mape_wg)
        print(f"Warengruppe {wg} MAPE: {mape_wg:.4%}")   #shows as percentage
    


    return  y_pred, metrics, warengruppes, warengruppes_mape

# Evaluate on Train and Validation sets
train_predictions, train_metrics, train_wg, train_wg_mape = evaluate_model(model, train_df, "Training Set")
val_predictions, val_metrics, val_wg, val_wg_mape = evaluate_model(model, val_df, "Validation Set")
```

Container dataframes.

```python
#make a dataframe with warengruppes and their mape for train and val sets
wg_performance_df = pd.DataFrame({
    'Warengruppe': train_wg,
    'Train_MAPE': train_wg_mape,
    'Val_MAPE': val_wg_mape
})

# build a dataframe with overall metrics
overall_metrics_df = pd.DataFrame({
    'Metric': ['R2_Score', 'MAE', 'MSE', 'RMSE', 'MAPE'],
    'Train': train_metrics,
    'Validation': val_metrics
})
```

Plotting the global metrics.

```python
# melt the overall metrics dataframe for better visualization
overall_metrics_melted = overall_metrics_df.melt(id_vars='Metric', var_name='Dataset', value_name='Value')

# Set color palette
sns.set_theme(style="whitegrid", context="talk", font_scale=1.05)
palette = sns.color_palette("viridis", 2) # Distinct colors for Train/Val

# --- Plot 1: Global Metrics (Subplots due to scale) ---
fig1, axes = plt.subplots(2, 2, figsize=(14, 10))
fig1.suptitle('Global Metrics for Model Performance', fontsize=24, weight='bold', y=1.02)

# We will plot R2, MAE, RMSE, MAPE (skipping MSE to save space/redundancy with RMSE, 
# and because MSE is less interpretable in original units)
metrics_to_plot = ['R2_Score', 'MAE', 'RMSE', 'MAPE']
axes_flat = axes.flatten()

for i, metric in enumerate(metrics_to_plot):
    ax = axes_flat[i]
    subset = overall_metrics_melted [overall_metrics_melted ['Metric'] == metric]
    
    # Seaborn Barplot
    sns.barplot(data=subset, x='Dataset', y='Value', hue='Dataset', palette=palette, ax=ax, dodge=False)
    # Customize y-axis format for MAPE as percentage
    if metric == 'MAPE':
        ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: '{:.0%}'.format(y)))   
    
    ax.set_title(metric, pad=10)
    ax.set_xlabel('')
    ax.set_ylabel('')
        

    # Labeling
    for container in ax.containers:
        # Custom format based on metric
        fmt = '%.4f' if metric == 'R2_Score' else '%.2f'
        if metric == 'MAPE':
            fmt = '{:.4%}' # Percentage format
        ax.bar_label(container, fmt=fmt, padding=-10, fontsize=14, weight='bold', label_type='center')

#
plt.show()
```
